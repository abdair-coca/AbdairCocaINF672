```python
from sklearn.datasets import load_iris
import numpy as np

iris = load_iris()
X = iris.data      # matriz de características
y = iris.target    # etiquetas: 0, 1, 2
print(X.shape)
print(iris.feature_names)
```

```python
x0 = X[0]          # primera muestra: un vector
print(x0)
print(x0.shape)
```

```python
# fila = muestra ; columna = característica
col_2 = X[:, 2]    # 3ª característica de TODAS las muestras
print(col_2[:5])
print(col_2.shape)
```

```python
print(X.shape)     # (150, 4)
print(X.T.shape)   # transpuesta: (4, 150)
```

```python
a = X[0]           # una flor setosa
b = X[50]          # una flor versicolor
print('a + b =', a + b)
print('2 * a =', 2 * a)
```

```python
manual = sum(ai * bi for ai, bi in zip(a, b))
print('a mano :', round(manual, 2))
print('np.dot :', round(np.dot(a, b), 2))
print('con @  :', round(a @ b, 2))
```

```python
# ADVERTENCIA: a * b NO es producto punto, es multiplicación elemento a elemento
print('a * b (elemento a elemento) =', a * b)
```

```python
def coseno(u, v):
    return np.dot(u, v) / (np.linalg.norm(u) * np.linalg.norm(v))

print('cos(a, b) =', round(coseno(a, b), 4))
```

```python
setosa1    = X[0]
setosa2    = X[1]
versicolor = X[50]
print('cos(setosa1, setosa2)    =', round(coseno(setosa1, setosa2), 4))
print('cos(setosa1, versicolor) =', round(coseno(setosa1, versicolor), 4))
```

```python
def distancia(u, v):
    return np.linalg.norm(u - v)

print('d(setosa1, setosa2)    =', round(distancia(setosa1, setosa2), 4))
print('d(setosa1, versicolor) =', round(distancia(setosa1, versicolor), 4))
```

```python
x = X[0]
print('x . x      =', round(np.dot(x, x), 4))
print('||x|| ** 2 =', round(np.linalg.norm(x) ** 2, 4))
```

## 07 · Entregable evaluable

Resuelto con NumPy (solo `np.dot` y `np.linalg.norm`); sin funciones de distancia/similitud de scikit-learn.

```python
# 1) Cargar iris y seleccionar 3 muestras:
#    i = 0  -> setosa     (y[0]  = 0)
#    j = 1  -> setosa     (y[1]  = 0)  (misma especie que i)
#    k = 50 -> versicolor (y[50] = 1)  (especie distinta)
idx = [0, 1, 50]
muestras = X[idx]
print('Indices     :', idx)
print('Etiquetas   :', y[idx])
print('Matriz 3x4 de muestras:\n', muestras)
```

```python
# 2) Matriz 3x3 de distancias euclidianas
D = np.zeros((3, 3))
for i in range(3):
    for j in range(3):
        D[i, j] = distancia(muestras[i], muestras[j])
print('Matriz de distancias euclidianas 3x3:\n', D)
```

```python
# 3) Matriz 3x3 de similitud coseno
C = np.zeros((3, 3))
for i in range(3):
    for j in range(3):
        C[i, j] = coseno(muestras[i], muestras[j])
print('Matriz de similitud coseno 3x3:\n', C)
```

```python
# 4) Verificar para muestras[0] que ||x||^2 = x . x
xv = muestras[0]
print('x . x      =', round(np.dot(xv, xv), 4))
print('||x|| ** 2 =', round(np.linalg.norm(xv) ** 2, 4))
print('diferencia =', round(np.dot(xv, xv) - np.linalg.norm(xv) ** 2, 12))
```

**Interpretación (3–4 líneas).**  
El par más parecido según la similitud coseno es `(setosa1, setosa2)` con coseno ≈ 0.9984, y también es el más cercano según la distancia euclidiana (0.5385). Ambas métricas coinciden porque ambas muestras son de la misma especie, por lo que sus vectores apuntan en direcciones casi idénticas y están cerca en el espacio. La distancia euclidiana depende también de la magnitud del vector, mientras que el coseno solo de la dirección; por eso pueden diferir cuando cambia el tamaño pero se mantiene la forma.
