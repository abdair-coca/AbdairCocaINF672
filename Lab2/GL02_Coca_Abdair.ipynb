<p style="font-size:14.5px;color:#334155">Primero cargamos las cuatro bibliotecas que acompañan toda la materia. Los alias <code>np</code>, <code>pd</code>, <code>plt</code> y <code>sns</code> son la convención universal: los encontrarás en la documentación oficial y en toda la bibliografía.</p>

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

<h3 style="color:#1E293B;margin:14px 0 4px;font-size:17px">Crear arreglos</h3><p style="font-size:14px;color:#334155;margin:6px 0">El <b>ndarray</b> es una colección de elementos del mismo tipo, organizados en una o más dimensiones. A diferencia de una lista de Python, permite operaciones <b>vectorizadas</b>: aplicar una operación a todos los elementos sin escribir bucles.</p><div style="border-left:5px solid #3B82F6;background:#EFF6FF;border-radius:8px;padding:12px 16px;margin:10px 0"><span style="color:#3B82F6;font-weight:800;font-size:12px;letter-spacing:2px">DEFINICIÓN</span><div style="font-weight:700;margin:2px 0">ndarray</div><div style="font-size:14px;color:#334155">El arreglo n-dimensional es la estructura central de NumPy: más rápido y conciso que las listas, porque ejecuta las operaciones en código optimizado por debajo.</div></div>

```python
a = np.array([2, 4, 6, 8])
a
```

<p style="font-size:14px;color:#334155;margin:6px 0">Todo arreglo conoce su <b>forma</b>, su <b>tipo de dato</b> y su número de <b>dimensiones</b>:</p>

```python
a.shape, a.dtype, a.ndim
```

<h3 style="color:#1E293B;margin:14px 0 4px;font-size:17px">Operaciones vectorizadas</h3><p style="font-size:14px;color:#334155;margin:6px 0">La operación se aplica elemento a elemento sobre todo el arreglo, sin bucles:</p>

```python
a * 10
```

<p style="font-size:14px;color:#334155;margin:6px 0">Un arreglo 2D es una <b>matriz</b> de filas y columnas. Con <code>axis</code> elegimos la dirección en que agrega la función:</p>

```python
M = np.array([[1, 2, 3],
              [4, 5, 6]])
M.shape, M.dtype, M.ndim
```

<p style="font-size:14px;color:#334155;margin:6px 0"><code>axis=0</code> recorre las filas y suma cada <b>columna</b>:</p>

```python
M.sum(axis=0)
```

<div style="border-left:5px solid #EF4444;background:#FEF2F2;border-radius:8px;padding:12px 16px;margin:10px 0"><span style="color:#EF4444;font-weight:800;font-size:12px;letter-spacing:2px">CLAVE</span><div style="font-weight:700;margin:2px 0">¿Por qué vectorizar?</div><div style="font-size:14px;color:#334155">No es solo más corto: NumPy ejecuta estas operaciones en código compilado y optimizado. Sobre miles de datos — como en Machine Learning — esa diferencia es decisiva.</div></div>

<h3 style="color:#1E293B;margin:14px 0 4px;font-size:17px">Del arreglo al dataset real: iris</h3><p style="font-size:14px;color:#334155;margin:6px 0">Trabajaremos con <b>iris</b>, un dataset clásico de 150 flores descritas por 4 medidas (largo y ancho de sépalo y pétalo). Viene incluido en <code>scikit-learn</code> y sus datos son, precisamente, un arreglo de NumPy:</p>

```python
from sklearn.datasets import load_iris
iris = load_iris()
iris.data.shape
```

<p style="font-size:14px;color:#334155;margin:6px 0">Agregación por <b>columnas</b>: la media de cada característica. <code>iris.data</code> es un ndarray puro: rápido para calcular, pero sin nombres ni etiquetas legibles.</p>

```python
iris.data.mean(axis=0)
```

<div style="border-left:5px solid #F59E0B;background:#FFFBEB;border-radius:8px;padding:12px 16px;margin:10px 0"><span style="color:#F59E0B;font-weight:800;font-size:12px;letter-spacing:2px">NOTA</span><div style="font-weight:700;margin:2px 0">El salto a Pandas</div><div style="font-size:14px;color:#334155">El ndarray es ideal para el cálculo numérico, pero no sabe qué significa cada columna. Ese es el trabajo de Pandas, que viene a continuación.</div></div>

<div style="background:linear-gradient(90deg,#0EA5E9 0%,#06B6D4 100%);color:white;padding:12px 20px;border-radius:10px;font-size:20px;font-weight:800;margin:16px 0 8px;display:flex;align-items:center;gap:10px"><span style="background:rgba(255,255,255,.22);border-radius:8px;padding:2px 10px;font-size:14px;font-weight:700">02</span>Pandas — Series</div>

<p style="font-size:14px;color:#334155">Una <b>Series</b> es un arreglo unidimensional <i>etiquetado</i>: como una columna de una tabla, donde cada valor tiene asociada una etiqueta (su índice). Combina la velocidad de NumPy con un índice que da significado a cada dato.</p><div style="border-left:5px solid #3B82F6;background:#EFF6FF;border-radius:8px;padding:12px 16px;margin:10px 0"><span style="color:#3B82F6;font-weight:800;font-size:12px;letter-spacing:2px">DEFINICIÓN</span><div style="font-weight:700;margin:2px 0">Series</div><div style="font-size:14px;color:#334155">Un vector 1D etiquetado. Cada dato posee un índice que lo identifica.</div></div>

```python
s = pd.Series([5.1, 4.9, 4.7, 4.6],
              index=["m1", "m2", "m3", "m4"])
s
```

<p style="font-size:14px;color:#334155;margin:6px 0">Se accede por <b>etiqueta</b> (con corchetes) o por <b>posición</b> (con <code>.iloc</code>):</p>

```python
s["m1"], s.iloc[0]
```

<p style="font-size:14px;color:#334155;margin:6px 0">El <b>filtrado booleano</b> selecciona los valores que cumplen una condición, conservando sus etiquetas:</p>

```python
s[s > 4.7]
```

<p style="font-size:14px;color:#334155;margin:6px 0">Los métodos estadísticos resumen la serie en un instante:</p>

```python
s.mean(), s.max()
```

<div style="background:linear-gradient(90deg,#10B981 0%,#14B8A6 100%);color:white;padding:12px 20px;border-radius:10px;font-size:20px;font-weight:800;margin:16px 0 8px;display:flex;align-items:center;gap:10px"><span style="background:rgba(255,255,255,.22);border-radius:8px;padding:2px 10px;font-size:14px;font-weight:700">03</span>Pandas — DataFrames con iris</div>

<p style="font-size:14px;color:#334155">Un <b>DataFrame</b> es una tabla bidimensional: un conjunto de columnas (cada una es una Series) que comparten el mismo índice de filas. Es la estructura que más usarás: representa un dataset completo, con filas (muestras) y columnas (características).</p><div style="border-left:5px solid #3B82F6;background:#EFF6FF;border-radius:8px;padding:12px 16px;margin:10px 0"><span style="color:#3B82F6;font-weight:800;font-size:12px;letter-spacing:2px">DEFINICIÓN</span><div style="font-weight:700;margin:2px 0">DataFrame</div><div style="font-size:14px;color:#334155">La estructura de datos central de Pandas: representa un dataset completo en forma de tabla.</div></div>

<p style="font-size:14px;color:#334155;margin:6px 0">Construimos el DataFrame de iris con nombres de columna cortos y agregamos la <b>especie</b> como columna categórica:</p>

```python
df = pd.DataFrame(iris.data,
                  columns=["sepal_len", "sepal_wid",
                           "petal_len", "petal_wid"])
df["species"] = pd.Categorical.from_codes(iris.target,
                                            iris.target_names)
df.head()
```

<p style="font-size:14px;color:#334155;margin:6px 0">Antes de analizar conviene mirar la <b>forma</b> y un <b>resumen estadístico</b>:</p>

```python
df.shape
```

```python
df["sepal_len"].describe()
```

<h3 style="color:#1E293B;margin:14px 0 4px;font-size:17px">Seleccionar y filtrar</h3><p style="font-size:14px;color:#334155;margin:6px 0">Se selecciona una columna con corchetes y se filtran filas con una condición booleana, igual que en NumPy pero conservando etiquetas:</p>

```python
df[df["species"] == "setosa"].head(3)
```

<h3 style="color:#1E293B;margin:14px 0 4px;font-size:17px">Columnas derivadas y groupby</h3><p style="font-size:14px;color:#334155;margin:6px 0">Creamos una columna nueva a partir de las existentes. Aquí, el <b>ratio pétalo</b> (largo entre ancho):</p>

```python
df["petal_ratio"] = df["petal_len"] / df["petal_wid"]
df[["petal_len", "petal_wid", "petal_ratio"]].head(3)
```

<p style="font-size:14px;color:#334155;margin:6px 0"><code>groupby</code> agrupa por una categoría y resume cada grupo. Es una de las operaciones más potentes de Pandas:</p>

```python
df.groupby("species")[["sepal_len", "petal_len"]].mean()
```

<div style="border-left:5px solid #8B5CF6;background:#F5F3FF;border-radius:8px;padding:12px 16px;margin:10px 0"><span style="color:#8B5CF6;font-weight:800;font-size:12px;letter-spacing:2px">EJEMPLO</span><div style="font-weight:700;margin:2px 0">Patrón clave</div><div style="font-size:14px;color:#334155">setosa tiene pétalos mucho más pequeños que versicolor y virginica. Ese tipo de diferencia entre grupos es lo que más adelante permitirá a un modelo distinguir las clases.</div></div>

<h3 style="color:#1E293B;margin:14px 0 4px;font-size:17px">Cargar datos desde un archivo (read_csv)</h3><p style="font-size:14px;color:#334155;margin:6px 0">En la práctica los datos casi nunca vienen de una librería: llegan en archivos. <code>read_csv</code> es la puerta de entrada de la mayoría de los proyectos. Guardemos iris a CSV y leámoslo de vuelta:</p>

```python
df.to_csv("iris.csv", index=False)
datos = pd.read_csv("iris.csv")
datos.head(2)
```

<div style="border-left:5px solid #F59E0B;background:#FFFBEB;border-radius:8px;padding:12px 16px;margin:10px 0"><span style="color:#F59E0B;font-weight:800;font-size:12px;letter-spacing:2px">NOTA</span><div style="font-weight:700;margin:2px 0">Parámetros útiles de read_csv</div><div style="font-size:14px;color:#334155"><code>sep</code> para el separador, <code>header</code> y <code>names</code> para los encabezados, <code>index_col</code> para elegir la columna índice y <code>na_values</code> para marcar datos faltantes. Volverás a él en casi todas las prácticas.</div></div>

<div style="background:linear-gradient(90deg,#F59E0B 0%,#F97316 100%);color:white;padding:12px 20px;border-radius:10px;font-size:20px;font-weight:800;margin:16px 0 8px;display:flex;align-items:center;gap:10px"><span style="background:rgba(255,255,255,.22);border-radius:8px;padding:2px 10px;font-size:14px;font-weight:700">04</span>Matplotlib — visualización</div>

<p style="font-size:14px;color:#334155">Matplotlib es la librería base de gráficos en Python. El patrón habitual: <b>dibujar</b> con una función (<code>plt.scatter</code>, <code>plt.hist</code>…), <b>etiquetar</b> ejes y título, y cerrar con <code>plt.show()</code>.</p>

<h3 style="color:#1E293B;margin:14px 0 4px;font-size:17px">Dispersión (scatter)</h3><p style="font-size:14px;color:#334155;margin:6px 0">Relaciona dos variables numéricas: el gráfico más usado para ver relaciones entre características.</p>

```python
plt.scatter(df["sepal_len"], df["petal_len"])
plt.xlabel("Largo de sépalo (cm)")
plt.ylabel("Largo de pétalo (cm)")
plt.title("Iris: sépalo vs pétalo")
plt.show()
```

<h3 style="color:#1E293B;margin:14px 0 4px;font-size:17px">Histograma</h3><p style="font-size:14px;color:#334155;margin:6px 0">Muestra cómo se distribuye una sola variable, agrupando los valores en intervalos (<code>bins</code>):</p>

```python
plt.hist(df["sepal_len"], bins=15, color="#4F46E5",
         edgecolor="white")
plt.xlabel("Largo de sépalo (cm)")
plt.ylabel("Frecuencia")
plt.title("Distribución del largo de sépalo")
plt.show()
```

<h3 style="color:#1E293B;margin:14px 0 4px;font-size:17px">Barras (con Pandas)</h3><p style="font-size:14px;color:#334155;margin:6px 0">Pandas usa Matplotlib por debajo: el método <code>.plot()</code> de una Series genera el gráfico directamente. Aquí, el largo de pétalo medio por especie:</p>

```python
medias = df.groupby("species")["petal_len"].mean()
medias.plot(kind="bar")
plt.ylabel("Largo de pétalo medio (cm)")
plt.title("Largo de pétalo medio por especie")
plt.show()
```

<h3 style="color:#1E293B;margin:14px 0 4px;font-size:17px">Varios ejes (subplots)</h3><p style="font-size:14px;color:#334155;margin:6px 0"><code>plt.subplots</code> crea una figura con varios ejes para comparar gráficos lado a lado. Cada eje se controla por separado:</p>

```python
fig, axes = plt.subplots(1, 2, figsize=(9, 3))
axes[0].scatter(df["sepal_len"], df["sepal_wid"])
axes[0].set_title("Sépalo")
axes[1].scatter(df["petal_len"], df["petal_wid"])
axes[1].set_title("Pétalo")
plt.show()
```

<div style="background:linear-gradient(90deg,#EC4899 0%,#F43F5E 100%);color:white;padding:12px 20px;border-radius:10px;font-size:20px;font-weight:800;margin:16px 0 8px;display:flex;align-items:center;gap:10px"><span style="background:rgba(255,255,255,.22);border-radius:8px;padding:2px 10px;font-size:14px;font-weight:700">05</span>Seaborn — exploración rápida (EDA)</div>

<p style="font-size:14px;color:#334155">Seaborn se construye sobre Matplotlib y trabaja directamente con DataFrames. Con muy poco código produce gráficos estadísticos vistosos, y el parámetro <code>hue</code> colorea automáticamente por categoría. Es la herramienta ideal para el <b>análisis exploratorio (EDA)</b>.</p>

<h3 style="color:#1E293B;margin:14px 0 4px;font-size:17px">Boxplot y dispersión por categoría</h3><p style="font-size:14px;color:#334155;margin:6px 0">Un <b>boxplot</b> compara la distribución de una variable entre grupos. Un <b>scatter con hue</b> separa las clases por color:</p>

```python
sns.boxplot(data=df, x="species", y="petal_len")
plt.title("Largo de pétalo por especie")
plt.show()
```

```python
sns.scatterplot(data=df, x="sepal_len", y="petal_len",
                hue="species")
plt.title("Iris: sépalo vs pétalo por especie")
plt.show()
```

<h3 style="color:#1E293B;margin:14px 0 4px;font-size:17px">Mapa de correlación</h3><p style="font-size:14px;color:#334155;margin:6px 0">El <b>heatmap de correlación</b> resume, de un vistazo, qué variables se mueven juntas. Es de los gráficos más usados al comenzar a explorar un dataset:</p>

```python
corr = df.drop(columns="species").corr()
sns.heatmap(corr, annot=True, cmap="viridis")
plt.title("Matriz de correlación de iris")
plt.show()
```

<div style="border-left:5px solid #10B981;background:#ECFDF5;border-radius:8px;padding:12px 16px;margin:10px 0"><span style="color:#10B981;font-weight:800;font-size:12px;letter-spacing:2px">TRUCO</span><div style="font-weight:700;margin:2px 0">Exploración completa en una sola línea</div><div style="font-size:14px;color:#334155"><code>sns.pairplot(df, hue="species")</code> dibuja todas las combinaciones de variables a la vez. Es más pesado de calcular, pero muy revelador.</div></div>

```python
sns.pairplot(df, hue="species")
```
