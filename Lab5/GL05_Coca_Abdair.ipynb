Los recuadros <b>In [ ]</b> son código; los <b>Out[ ]</b> la salida esperada.</p><h3 style="color:#1E293B;margin:14px 0 4px;font-size:17px">4.0 Preparación del entorno</h3><p style="font-size:14px;color:#334155;margin:6px 0">Importamos librerías, fijamos paleta INF672 y cargamos iris en DataFrame.</p>

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris

# --- paleta INF672 (para que los gráficos combinen con el documento) ---
INDIGO, PURPLE, CORAL, INK, GREY = "#4F46E5", "#7E22CE", "#F43F5E", "#111827", "#6B7280"
plt.rcParams.update({
    "axes.edgecolor": "#D1D5DB", "axes.titleweight": "bold", "axes.titlecolor": INK,
    "axes.labelcolor": INK, "text.color": INK, "xtick.color": INK, "ytick.color": INK,
    "axes.grid": True, "grid.color": "#E5E7EB", "grid.linewidth": 0.7, "axes.axisbelow": True,
    "figure.dpi": 110,
})

iris = load_iris()
df = pd.DataFrame(iris.data, columns=["sep_largo", "sep_ancho", "pet_largo", "pet_ancho"])
print("Forma del dataset:", df.shape)
df.head()
```

<h3 style="color:#1E293B;margin:14px 0 4px;font-size:17px">4.1 Tendencia central: media, mediana y moda</h3>

```python
col = df["pet_largo"]
print("media   :", round(col.mean(), 3))
print("mediana :", round(col.median(), 3))
print("moda    :", col.mode().tolist())  # la moda es más útil en variables discretas/categóricas
```

<div style="background:#FFFBEB;border:1px solid #FDE68A;border-radius:10px;padding:14px 18px;font-size:14px;color:#334155"><b>Out[2] esperado:</b> <code>media: 3.758 · mediana: 4.35 · moda: [1.4, 1.5]</code><br>La media (3.76) y la mediana (4.35) <b>no coinciden</b>: los datos no son simétricos. Un grupo de flores muy pequeñas —la especie <i>setosa</i>— jala la media hacia abajo. La bimodalidad (ver §4.3) explica este desacople.</div>

<div style="background:#F5F3FF;border-left:5px solid #7C3AED;border-radius:8px;padding:12px 16px;margin:10px 0"><span style="color:#7C3AED;font-weight:800;font-size:12px;letter-spacing:2px">LA IDEA QUE IMPORTA</span><div style="font-weight:700;margin:2px 0">La mediana es robusta</div><div style="font-size:14px;color:#334155">Inyectamos un valor absurdo en el ancho del sépalo y observamos qué le pasa a cada medida.</div></div>

```python
base = df["sep_ancho"].values
conout = np.append(base, 12.0)  # un ancho de sépalo imposible

fig, axes = plt.subplots(1, 2, figsize=(9, 3.2))
for ax, data, titulo in [(axes[0], base, "Sin outlier"),
                         (axes[1], conout, "Con un outlier (12.0)")]:
    ax.hist(data, bins=18, color=INDIGO, alpha=0.7, edgecolor="white", linewidth=0.5)
    ax.axvline(data.mean(), color=CORAL, lw=2.2, label=f"media={data.mean():.2f}")
    ax.axvline(np.median(data), color=PURPLE, lw=2.2, ls="--", label=f"mediana={np.median(data):.2f}")
    ax.set_title(titulo); ax.set_xlabel("Ancho del sépalo (cm)"); ax.legend(fontsize=9)
axes[0].set_ylabel("Frecuencia")
fig.suptitle("Un solo outlier arrastra la media, no la mediana", fontweight="bold")
plt.tight_layout(); plt.show()
```

<div style="background:#FEF2F2;border:1px solid #FECACA;border-radius:10px;padding:14px 18px;font-size:14px;color:#334155"><b>Lectura:</b> un único valor extremo desplaza la <span style="color:#F43F5E;font-weight:700">media</span> pero deja casi intacta la <span style="color:#7E22CE;font-weight:700">mediana</span> (≈3.00 sin y con outlier). Esa insensibilidad es la <b>robustez</b>.</div><div style="border-left:5px solid #8B5CF6;background:#F5F3FF;border-radius:8px;padding:12px 16px;margin:10px 0"><span style="color:#8B5CF6;font-weight:800;font-size:12px;letter-spacing:2px">CONEXIÓN ML</span><div style="font-size:14px;color:#334155">Al rellenar faltantes (tema 2.3.1) elegirás media o mediana según haya outliers: la <b>mediana es la opción segura</b>. Misma robustez base para detección de anomalías.</div></div>

<h3 style="color:#1E293B;margin:14px 0 4px;font-size:17px">4.4 Relación entre dos variables: correlación</h3><p style="font-size:14px;color:#334155;margin:6px 0">Pearson normaliza covarianza a [−1,+1]. Con una llamada obtenemos matriz de todas las parejas.</p>

```python
df.corr().round(2)
```

<div style="background:#F8FAFC;border:1px solid #E2E8F0;border-radius:10px;padding:14px 18px;font-size:14px;color:#334155"><b>Lectura tabla:</b> sep_largo-pet_largo 0.87 fuerte positiva; pet_largo-pet_ancho 0.96 casi perfecta; sep_ancho correlaciones negativas débiles (−0.12 a −0.43).</div>

```python
C = df.corr().values
labels = ["sépalo\nlargo", "sépalo\nancho", "pétalo\nlargo", "pétalo\nancho"]
fig, ax = plt.subplots(figsize=(5, 4.2))
im = ax.imshow(C, cmap="PuOr", vmin=-1, vmax=1)
ax.set_xticks(range(4), labels, fontsize=9); ax.set_yticks(range(4), labels, fontsize=9)
for i in range(4):
    for j in range(4):
        ax.text(j, i, f"{C[i,j]:.2f}", ha="center", va="center",
                color="white" if abs(C[i,j]) > 0.55 else INK, fontweight="bold")
ax.set_title("Matriz de correlación (iris)"); ax.grid(False)
fig.colorbar(im, fraction=0.046, pad=0.04); plt.tight_layout(); plt.show()
```

```python
x, y = df["pet_largo"].values, df["pet_ancho"].values
r = np.corrcoef(x, y)[0, 1]
b1, b0 = np.polyfit(x, y, 1)
plt.figure(figsize=(6, 3.6))
plt.scatter(x, y, color=INDIGO, alpha=0.6, edgecolor="white", linewidth=0.4, s=38)
plt.plot([x.min(), x.max()], b0 + b1*np.array([x.min(), x.max()]), color=CORAL, lw=2)
plt.xlabel("Largo del pétalo (cm)"); plt.ylabel("Ancho del pétalo (cm)")
plt.title(f"Dos características muy correlacionadas (r = {r:.2f})")
plt.tight_layout(); plt.show()
```

<div style="background:#F8FAFC;border:1px solid #E2E8F0;border-radius:10px;padding:14px 18px;font-size:14px;color:#334155">Par más correlacionado: pet_largo-pet_ancho r=0.96. Scatter con recta de ajuste <code>np.polyfit</code> muestra nube alargada muy lineal.</div>

<div style="border-left:5px solid #4F46E5;background:#EFF6FF;border-radius:8px;padding:12px 16px;margin:10px 0"><span style="color:#4F46E5;font-weight:800;font-size:12px;letter-spacing:2px">EL PUENTE CON EL ÁLGEBRA LINEAL (2.1.1)</span><div style="font-size:14px;color:#334155;margin-top:4px">Pearson es exactamente el <b>coseno del ángulo entre vectores centrados</b> (a cada variable se le resta su media). Misma similitud coseno del producto punto.</div></div>

```python
xc, yc = x - x.mean(), y - y.mean()
cos = xc @ yc / (np.linalg.norm(xc) * np.linalg.norm(yc))
print("Pearson (np.corrcoef) :", round(np.corrcoef(x, y)[0, 1], 6))
print("coseno centrado       :", round(float(cos), 6))
print("¿idénticos? ->", np.isclose(np.corrcoef(x, y)[0, 1], cos))
```

<div style="background:#ECFDF5;border:1px solid #A7F3D0;border-radius:10px;padding:14px 18px;font-size:14px;color:#334155"><b>Out[9]:</b> ambos 0.962865, <code>True</code>. Fórmula: <code>corr = (x−x̄)·(y−ȳ) / (‖x−x̄‖·‖y−ȳ‖)</code>.</div><div style="border-left:5px solid #8B5CF6;background:#F5F3FF;border-radius:8px;padding:12px 16px;margin:10px 0"><span style="color:#8B5CF6;font-weight:800;font-size:12px;letter-spacing:2px">CONEXIÓN ML</span><div style="font-size:14px;color:#334155">Dos características r=0.96 cargan casi misma información: una es redundante. Detectarlo guía <b>selección de características (2.4)</b> y previene multicolinealidad en modelos lineales.</div></div>

En pet_largo Δ≈0.59 por bimodalidad. <b>Decisión ML:</b> imputar faltantes en sep_largo con media o mediana da similar; en pet_largo conviene mediana.</div>

<h3 style="color:#1E293B;margin:14px 0 4px;font-size:17px">Ej. 2 — Robustez sobre pet_ancho con outlier extremo</h3><p style="font-size:14px;color:#334155;margin:6px 0">Replicamos el experimento de robustez inyectando 5.0 cm (el doble del máximo real 2.5) en pet_ancho.</p>

```python
base2 = df["pet_ancho"].values
conout2 = np.append(base2, 5.0)
print(f"Sin outlier — media {base2.mean():.3f}  mediana {np.median(base2):.3f}")
print(f"Con 5.0     — media {conout2.mean():.3f}  mediana {np.median(conout2):.3f}")
print(f"Δ media {conout2.mean()-base2.mean():+.3f}  Δ mediana {np.median(conout2)-np.median(base2):+.3f}")

fig, axes = plt.subplots(1, 2, figsize=(9, 3.2))
for ax, data, ttl in [(axes[0], base2, "pet_ancho sin outlier"),(axes[1], conout2, "pet_ancho + outlier 5.0")]:
    ax.hist(data, bins=16, color=INDIGO, alpha=0.7, edgecolor="white", linewidth=0.5)
    ax.axvline(data.mean(), color=CORAL, lw=2.2, label=f"media={data.mean():.2f}")
    ax.axvline(np.median(data), color=PURPLE, lw=2.2, ls="--", label=f"mediana={np.median(data):.2f}")
    ax.set_title(ttl); ax.legend(fontsize=9)
plt.tight_layout(); plt.show()
```

<div style="background:#ECFDF5;border:1px solid #A7F3D0;border-radius:10px;padding:14px 18px;font-size:14px;color:#334155"><b>Interpretación:</b> media sube ~0.025 con outlier; mediana casi inmóvil. Confirma robustez en otra variable. <b>ML:</b> si pet_ancho tuviera errores de medición, imputar con mediana evita arrastrar el centro.</div>

<h3 style="color:#1E293B;margin:14px 0 4px;font-size:17px">Ej. 3 — Forma de distribución: sep_largo vs pet_largo</h3><p style="font-size:14px;color:#334155;margin:6px 0">Comparamos histogramas para contrastar unimodal simétrico vs bimodal.</p>

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 3.4))
for ax, col_name in [(axes[0], "sep_largo"), (axes[1], "pet_largo")]:
    v = df[col_name].values
    ax.hist(v, bins=18, color=INDIGO, alpha=0.75, edgecolor="white", linewidth=0.6)
    ax.axvline(v.mean(), color=CORAL, lw=2.2, label=f"media={v.mean():.2f}")
    ax.axvline(np.median(v), color=PURPLE, lw=2.2, ls="--", label=f"mediana={np.median(v):.2f}")
    ax.set_title(f"{col_name} — {'unimodal' if col_name=='sep_largo' else 'bimodal'}")
    ax.set_xlabel(f"{col_name} (cm)"); ax.legend()
axes[0].set_ylabel("Frecuencia")
plt.tight_layout(); plt.show()
```

<h3 style="color:#1E293B;margin:14px 0 4px;font-size:17px">Ej. 4 — Dispersión comparada y necesidad de escalado</h3><p style="font-size:14px;color:#334155;margin:6px 0">Cuantificamos qué característica domina por escala y simulamos efecto sobre distancia euclídea.</p>

```python
disp = df.std(ddof=0).sort_values(ascending=False)
print(disp.round(3))
print(f"\nRatio σ_max/σ_min = {disp.max()/disp.min():.2f}  (pet_largo {disp.max():.2f} vs sep_ancho {disp.min():.2f})")
# Distancia euclídea dominada por pet_largo sin escalado
from sklearn.preprocessing import StandardScaler
a, b = df.iloc[0].values, df.iloc[50].values  # setosa vs versicolor
d_raw = np.linalg.norm(a-b)
scaler = StandardScaler().fit(df.values)
d_scaled = np.linalg.norm(scaler.transform([a])[0] - scaler.transform([b])[0])
print(f"Distancia setosa-versicolor raw: {d_raw:.2f}  escalada: {d_scaled:.2f}")
```

<div style="background:#ECFDF5;border:1px solid #A7F3D0;border-radius:10px;padding:14px 18px;font-size:14px;color:#334155"><b>Interpretación:</b> σ pet_largo (1.76) ≈4× σ sep_ancho (0.43). Sin escalado KNN/K-means es dominado por pétalo. Tras StandardScaler la distancia refleja aporte equilibrado → justifica <b>escalado 2.3.3</b>.</div>

<h3 style="color:#1E293B;margin:14px 0 4px;font-size:17px">Ej. 5 — Correlación débil: sep_largo vs sep_ancho (r≈−0.12)</h3><p style="font-size:14px;color:#334155;margin:6px 0">Contrastamos con el par muy correlacionado de la guía; visualizamos que r≈0 no implica nube sin forma sino relación lineal casi nula.</p>

```python
x2, y2 = df["sep_largo"].values, df["sep_ancho"].values
r2 = np.corrcoef(x2, y2)[0,1]
b1b, b0b = np.polyfit(x2, y2, 1)
plt.figure(figsize=(6, 3.6))
plt.scatter(x2, y2, color=INDIGO, alpha=0.6, edgecolor="white", linewidth=0.4, s=38)
plt.plot([x2.min(), x2.max()], b0b + b1b*np.array([x2.min(), x2.max()]), color=CORAL, lw=2)
plt.xlabel("Sépalo largo (cm)"); plt.ylabel("Sépalo ancho (cm)")
plt.title(f"Par poco correlacionado (r = {r2:.2f})")
plt.tight_layout(); plt.show()
print(f"r = {r2:.3f} — correlación lineal casi nula; pendiente {b1b:.3f}")
```

<div style="background:#ECFDF5;border:1px solid #A7F3D0;border-radius:10px;padding:14px 18px;font-size:14px;color:#334155"><b>Interpretación:</b> r=−0.12 indica relación lineal muy débil y negativa; puntos dispersos sin tendencia clara. <b>ML:</b> ambas variables aportan información casi ortogonal → ambas útiles, no redundantes, candidatas a conservarse en selección 2.4.</div>

<h3 style="color:#1E293B;margin:14px 0 4px;font-size:17px">Ej. 6 — Puente Pearson = coseno sobre otro par</h3><p style="font-size:14px;color:#334155;margin:6px 0">Verificamos identidad del puente para sep_largo-pet_largo (r=0.87) y mostramos que sin centrar el coseno difiere.</p>

```python
xa, ya = df["sep_largo"].values, df["pet_largo"].values
xc2, yc2 = xa - xa.mean(), ya - ya.mean()
cos2 = xc2 @ yc2 / (np.linalg.norm(xc2)*np.linalg.norm(yc2))
cos_sin_centrar = xa @ ya / (np.linalg.norm(xa)*np.linalg.norm(ya))
print(f"Pearson              : {np.corrcoef(xa, ya)[0,1]:.6f}")
print(f"coseno centrado      : {cos2:.6f}  -> isclose {np.isclose(np.corrcoef(xa,ya)[0,1], cos2)}")
print(f"coseno SIN centrar   : {cos_sin_centrar:.6f}  (distinto, demuestra que centrar importa)")
```

<div style="background:#ECFDF5;border:1px solid #A7F3D0;border-radius:10px;padding:14px 18px;font-size:14px;color:#334155"><b>Interpretación:</b> Pearson 0.8718 = coseno centrado; coseno sin centrar ≈0.98 sobrestima al no quitar la media (ángulo medido desde origen, no desde centro de nube). Confirma fórmula <code>corr = cos(x−x̄, y−ȳ)</code> y su invarianza a traslación, clave en similitud coseno.</div>
