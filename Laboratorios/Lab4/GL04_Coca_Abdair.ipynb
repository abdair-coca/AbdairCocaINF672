<div style="background:linear-gradient(90deg,#4F46E5 0%,#6366F1 100%);color:white;padding:12px 20px;border-radius:10px;font-size:20px;font-weight:800;margin:16px 0 8px;display:flex;align-items:center;gap:10px"><span style="background:rgba(255,255,255,.22);border-radius:8px;padding:2px 10px;font-size:14px;font-weight:700">01</span>Preparación del entorno</div><p style="font-size:14.5px;color:#334155">La primera celda importa librerías, guarda la paleta de la materia y fija un <b>estilo global</b> con <code>rcParams</code>. Así todas las figuras comparten la misma apariencia sin repetir configuración.</p>

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D  # habilita la proyección 3D

# Paleta INF672 — "cuaderno científico"
INDIGO = "#4F46E5"
PURPURA = "#7E22CE"
CORAL = "#F43F5E"
TINTA = "#111827"

plt.rcParams["figure.figsize"] = (8, 5)
plt.rcParams["axes.grid"] = True
plt.rcParams["grid.alpha"] = 0.3
plt.rcParams["font.size"] = 11
plt.rcParams["axes.edgecolor"] = "#9CA3AF"

print("Entorno listo ✓")
```

<div style="background:#F8FAFC;border:1px solid #E2E8F0;border-radius:10px;padding:14px 18px;font-size:14px;color:#334155"><b>Cómo se lee esta celda:</b><ul style="margin:8px 0 0 18px"><li><code>import numpy as np</code> — arreglos y matemática vectorizada: operar sobre miles de números sin bucles.</li><li><code>import matplotlib.pyplot as plt</code> — librería de dibujo, alias universal <code>plt</code>.</li><li><code>from mpl_toolkits.mplot3d import Axes3D</code> — no se usa por nombre, solo activa <code>projection="3d"</code>.</li><li>Constantes <code>INDIGO/CORAL</code> — identidad visual centralizada.</li><li><code>plt.rcParams</code> — diccionario global: tamaño figura, rejilla 30%, fuente, borde.</li></ul></div><div style="border-left:5px solid #4F46E5;background:#EFF6FF;border-radius:8px;padding:12px 16px;margin:10px 0"><span style="color:#4F46E5;font-weight:800;font-size:12px;letter-spacing:2px">CLAVE DE CÓDIGO</span><div style="font-weight:700;margin:2px 0">Configurar una sola vez, dibujar muchas veces.</div><div style="font-size:14px;color:#334155">Poner el estilo en <code>rcParams</code> evita repetir <code>figsize, grid</code> en cada celda.</div></div>

<div style="background:linear-gradient(90deg,#F59E0B 0%,#F97316 100%);color:white;padding:12px 20px;border-radius:10px;font-size:20px;font-weight:800;margin:16px 0 8px;display:flex;align-items:center;gap:10px"><span style="background:rgba(255,255,255,.22);border-radius:8px;padding:2px 10px;font-size:14px;font-weight:700">02</span>Visualización 1 · La recta tangente</div><p style="font-size:14px;color:#334155">Ya derivamos a mano que para <b>f(x)=x²</b> la pendiente es <b>f'(x)=2x</b>. El código no la recalcula: la recibe hecha y traza la recta tangente en <b>x₀=1.5</b> con la ecuación punto-pendiente <code>y = f(x₀)+f'(x₀)(x−x₀)</code>.</p>

```python
f = lambda x: x**2          # función
fp = lambda x: 2*x          # derivada obtenida A MANO: f'(x)=2x

x0 = 1.5
x = np.linspace(-3, 3, 400)
tangente = f(x0) + fp(x0) * (x - x0)  # recta: y = f(x0)+f'(x0)(x-x0)

plt.plot(x, f(x), color=INDIGO, lw=2.5, label=r"$f(x)=x^2$")
plt.plot(x, tangente, color=CORAL, lw=2, ls="--", label=fr"tangente en $x_0={x0}$")
plt.scatter([x0], [f(x0)], color=PURPURA, s=70, zorder=5)
plt.axhline(0, color="#9CA3AF", lw=0.8); plt.axvline(0, color="#9CA3AF", lw=0.8)
plt.title(f"Pendiente de la tangente = f'({x0}) = {fp(x0)}", color=TINTA)
plt.legend(); plt.show()
```

<div style="background:#FFFBEB;border:1px solid #FDE68A;border-radius:10px;padding:14px 18px;font-size:14px;color:#334155"><b>Construcción paso a paso:</b><ul style="margin:8px 0 0 18px"><li><code>f, fp</code> lambdas — <code>**</code> opera elemento a elemento, funcionan con escalar o arreglo.</li><li><code>np.linspace(-3,3,400)</code> — 400 puntos, más puntos = curva más suave.</li><li><code>tangente</code> vectorizada — se evalúa en los 400 puntos de golpe.</li><li><code>plt.plot(..., lw, label, r"$...$")</code> — <code>r"$"</code> activa LaTeX, <code>fr""</code> combina f-string + LaTeX.</li><li><code>plt.scatter(s=70, zorder=5)</code> — punto encima de líneas; <code>axhline/axvline</code> ejes grises.</li></ul></div><div style="border-left:5px solid #F59E0B;background:#FFFBEB;border-radius:8px;padding:12px 16px;margin:10px 0"><span style="color:#F59E0B;font-weight:800;font-size:12px;letter-spacing:2px">NOTA</span><div style="font-weight:700;margin:2px 0">Lectura del gráfico</div><div style="font-size:14px;color:#334155">La recta coral besa la parábola en un solo punto con inclinación 3.0 — el sentido geométrico de la derivada: pendiente instantánea.</div></div>

<div style="background:linear-gradient(90deg,#10B981 0%,#14B8A6 100%);color:white;padding:12px 20px;border-radius:10px;font-size:20px;font-weight:800;margin:16px 0 8px;display:flex;align-items:center;gap:10px"><span style="background:rgba(255,255,255,.22);border-radius:8px;padding:2px 10px;font-size:14px;font-weight:700">03</span>Visualización 2 · Superficie y curvas de nivel</div><p style="font-size:14px;color:#334155">Una función de dos variables <b>f(x,y)=x²+xy+y²</b> es una <b>superficie</b>. La dibujamos de dos formas complementarias: vista 3D (izq.) y mapa topográfico de curvas de nivel (der.).</p>

```python
f2 = lambda x, y: x**2 + x*y + y**2

x = np.linspace(-3, 3, 120)
y = np.linspace(-3, 3, 120)
X, Y = np.meshgrid(x, y)
Z = f2(X, Y)

fig = plt.figure(figsize=(12, 5))

ax1 = fig.add_subplot(1, 2, 1, projection="3d")
ax1.plot_surface(X, Y, Z, cmap="viridis", alpha=0.92)
ax1.set_title("Superficie f(x,y) = x² + xy + y²")
ax1.set_xlabel("x"); ax1.set_ylabel("y"); ax1.set_zlabel("f")

ax2 = fig.add_subplot(1, 2, 2)
cs = ax2.contourf(X, Y, Z, levels=20, cmap="viridis")
ax2.contour(X, Y, Z, levels=20, colors="white", linewidths=0.4)
ax2.set_title("Curvas de nivel"); ax2.set_xlabel("x"); ax2.set_ylabel("y")
ax2.set_aspect("equal")
fig.colorbar(cs, ax=ax2)
plt.tight_layout(); plt.show()
```

<div style="border-left:5px solid #3B82F6;background:#EFF6FF;border-radius:8px;padding:12px 16px;margin:10px 0"><span style="color:#3B82F6;font-weight:800;font-size:12px;letter-spacing:2px">DEFINICIÓN</span><div style="font-weight:700;margin:2px 0">np.meshgrid — la idea clave</div><div style="font-size:14px;color:#334155"><code>linspace</code> da ejes 1D (120 valores cada uno). <code>meshgrid</code> los combina en matrices 120×120 con coordenadas de cada casilla. <code>Z=f2(X,Y)</code> evalúa 14 400 puntos de una vez, sin bucles. Sin meshgrid no hay superficies.</div></div><div style="background:#F8FAFC;border:1px solid #E2E8F0;border-radius:10px;padding:14px 18px;font-size:14px;color:#334155"><b>Recorrido del código:</b><ul style="margin:8px 0 0 18px"><li><code>fig.add_subplot(1,2,1, projection="3d")</code> — 1 fila 2 col, casilla 1 en 3D.</li><li><code>plot_surface(cmap="viridis", alpha=0.92)</code> — colorea por altura (morado→amarillo).</li><li><code>contourf(levels=20)</code> — 20 bandas rellenas; <code>contour(colors="white")</code> líneas blancas encima.</li><li><code>set_aspect("equal")</code> — evita elipses deformadas; <code>colorbar</code> traduce color↔valor.</li></ul></div>

<div style="background:linear-gradient(90deg,#EC4899 0%,#F43F5E 100%);color:white;padding:12px 20px;border-radius:10px;font-size:20px;font-weight:800;margin:16px 0 8px;display:flex;align-items:center;gap:10px"><span style="background:rgba(255,255,255,.22);border-radius:8px;padding:2px 10px;font-size:14px;font-weight:700">04</span>Visualización 3 · El campo del gradiente</div><p style="font-size:14px;color:#334155">El gradiente reúne parciales en un vector: <b>∇f = (2x+y, x+2y)</b> (ya calculado a mano). Dibujamos una flecha en cada punto: hacia dónde y con qué fuerza crece f.</p>

```python
# Gradiente obtenido A MANO: ∇f = (2x+y, x+2y)
def grad_f2(x, y):
    return 2*x + y, x + 2*y

# Malla dispersa para las flechas
xg = np.linspace(-3, 3, 20)
yg = np.linspace(-3, 3, 20)
Xg, Yg = np.meshgrid(xg, yg)
U, V = grad_f2(Xg, Yg)

# Malla fina para los contornos de fondo
xf = np.linspace(-3, 3, 200); yf = np.linspace(-3, 3, 200)
Xf, Yf = np.meshgrid(xf, yf); Zf = f2(Xf, Yf)

plt.contour(Xf, Yf, Zf, levels=15, cmap="viridis", alpha=0.6)
plt.quiver(Xg, Yg, U, V, color=CORAL, alpha=0.85)
plt.title("Campo del gradiente ∇f (las flechas apuntan cuesta arriba)", color=TINTA)
plt.xlabel("x"); plt.ylabel("y"); plt.gca().set_aspect("equal")
plt.show()
```

<div style="background:#FEF2F2;border:1px solid #FECACA;border-radius:10px;padding:14px 18px;font-size:14px;color:#334155"><b>¿Por qué dos mallas?</b><ul style="margin:8px 0 0 18px"><li><code>grad_f2</code> devuelve tupla (U,V) — componentes del gradiente.</li><li>Dispersa 20×20 (400 flechas) — pocas y separadas; 200×200 serían mancha ilegible.</li><li>Fina 200×200 (40 000 pts) — curvas de nivel sí necesitan densidad para suavidad.</li><li><code>plt.quiver(Xg,Yg,U,V)</code> — coloca flecha (U,V) en base (Xg,Yg): campo de vectores.</li></ul></div><div style="border-left:5px solid #8B5CF6;background:#F5F3FF;border-radius:8px;padding:12px 16px;margin:10px 0"><span style="color:#8B5CF6;font-weight:800;font-size:12px;letter-spacing:2px">NOTA</span><div style="font-weight:700;margin:2px 0">Dos hechos que confirma el dibujo</div><div style="font-size:14px;color:#334155">(1) Flechas perpendiculares a curvas de nivel, (2) apuntan donde f crece. En la siguiente sección iremos en dirección <b>contraria</b> para bajar al mínimo.</div></div>

<div style="background:linear-gradient(90deg,#0EA5E9 0%,#06B6D4 100%);color:white;padding:12px 20px;border-radius:10px;font-size:20px;font-weight:800;margin:16px 0 8px;display:flex;align-items:center;gap:10px"><span style="background:rgba(255,255,255,.22);border-radius:8px;padding:2px 10px;font-size:14px;font-weight:700">05</span>Visualización 4 · Descenso de gradiente</div><p style="font-size:14px;color:#334155">Si ∇f apunta al máximo ascenso, <b>−∇f</b> apunta al descenso. El algoritmo da pasos pequeños en esa dirección hasta el mínimo. Usamos el tazón <b>f(x,y)=x²+y²</b>, ∇f=(2x,2y).</p><h3 style="color:#1E293B;margin:14px 0 4px;font-size:17px">Parte A · El bucle que calcula la trayectoria</h3>

```python
f3 = lambda x, y: x**2 + y**2
grad_f3 = lambda x, y: np.array([2*x, 2*y])  # ∇f=(2x,2y), derivado a mano

punto = np.array([4.0, 4.0])  # x_0
alpha = 0.1                   # tasa de aprendizaje
trayectoria = [punto.copy()]

for _ in range(25):
    punto = punto - alpha * grad_f3(*punto)  # x_{k+1}=x_k-α∇f(x_k)
    trayectoria.append(punto.copy())

trayectoria = np.array(trayectoria)
print("Primeros puntos:")
for i, p in enumerate(trayectoria[:3]):
    print(f" x_{i} = ({p[0]:.3f}, {p[1]:.3f})")
print(f" ...\n x_25 = ({trayectoria[-1,0]:.4f}, {trayectoria[-1,1]:.4f})")
```

<div style="background:#F8FAFC;border:1px solid #E2E8F0;border-radius:10px;padding:14px 18px;font-size:14px;color:#334155"><ul style="margin:0 0 0 18px"><li><code>grad_f3</code> devuelve <code>np.array</code> (no tupla) para aritmética vectorial directa.</li><li><code>punto=[4.0,4.0]</code> — <code>.0</code> fuerza float.</li><li><code>trayectoria=[punto.copy()]</code> — <code>.copy()</code> evita guardar referencia mutable (todos quedarían iguales).</li><li><code>grad_f3(*punto)</code> — <code>*</code> desempaqueta arreglo en args x,y.</li><li><code>:.3f / :.4f</code> — formato decimales; salida confirma descenso (4,4)→(0,0).</li></ul></div><div style="border-left:5px solid #F59E0B;background:#FFFBEB;border-radius:8px;padding:12px 16px;margin:10px 0"><span style="color:#F59E0B;font-weight:800;font-size:12px;letter-spacing:2px">NOTA</span><div style="font-size:14px;color:#334155"><b>Cuenta a mano:</b> x₁=(4,4)−0.1·(8,8)=(3.2,3.2), x₂=(2.56,2.56). El bucle repite esa resta 25 veces.</div></div>

<h3 style="color:#1E293B;margin:14px 0 4px;font-size:17px">Parte B · Dibujar la trayectoria sobre el mapa</h3>

```python
xf = np.linspace(-5, 5, 200); yf = np.linspace(-5, 5, 200)
Xf, Yf = np.meshgrid(xf, yf); Zf = f3(Xf, Yf)

plt.contour(Xf, Yf, Zf, levels=20, cmap="viridis", alpha=0.6)
plt.plot(trayectoria[:, 0], trayectoria[:, 1], "o-", color=CORAL, ms=4, lw=1.5, label="trayectoria")
plt.scatter([0], [0], marker="*", s=260, color=PURPURA, zorder=5, label="mínimo")
plt.title("Descenso de gradiente: bajando hacia el mínimo", color=TINTA)
plt.xlabel("x"); plt.ylabel("y"); plt.legend(); plt.gca().set_aspect("equal")
plt.show()
```

<div style="background:#F8FAFC;border:1px solid #E2E8F0;border-radius:10px;padding:14px 18px;font-size:14px;color:#334155"><ul style="margin:0 0 0 18px"><li><code>trayectoria[:,0]</code> y <code>[:,1]</code> — rebanado: columna x / y.</li><li><code>"o-"</code> — marcador circular + línea; <code>ms</code> tamaño marcador.</li><li><code>marker="*" s=260</code> — estrella mínimo en (0,0).</li><li><b>Espaciado:</b> puntos separados al inicio (gradiente grande, paso largo) y juntos cerca del mínimo (gradiente pequeño) — el algoritmo frena solo.</li></ul></div><div style="border-left:5px solid #EF4444;background:#FEF2F2;border-radius:8px;padding:12px 16px;margin:10px 0"><span style="color:#EF4444;font-weight:800;font-size:12px;letter-spacing:2px">CUIDADO</span><div style="font-weight:700;margin:2px 0">La tasa α es delicada</div><div style="font-size:14px;color:#334155">α muy grande → diverge/oscila. α muy pequeña → converge lento.</div></div>

<div style="background:linear-gradient(90deg,#8B5CF6 0%,#6366F1 100%);color:white;padding:12px 20px;border-radius:10px;font-size:20px;font-weight:800;margin:16px 0 8px;display:flex;align-items:center;gap:10px"><span style="background:rgba(255,255,255,.22);border-radius:8px;padding:2px 10px;font-size:14px;font-weight:700">06</span>Entregables — E2, E3 y E4</div>

<h3 style="color:#1E293B;margin:14px 0 4px;font-size:17px">E2 · Nueva tangente: g(x)=x³−3x en x₀=1</h3><p style="font-size:14px;color:#334155">Derivada a mano: <b>g'(x)=3x²−3</b>. En x₀=1 → g(1)=−2, g'(1)=0 ⇒ tangente horizontal <code>y=−2</code>. Repetimos la plantilla de Viz 1.</p><div style="border-left:5px solid #3B82F6;background:#EFF6FF;border-radius:8px;padding:12px 16px;margin:10px 0"><span style="color:#3B82F6;font-weight:800;font-size:12px;letter-spacing:2px">DERIVADA A MANO</span><div style="font-size:14px;color:#334155">g(x)=x³−3x → g'(x)=3x²−3 (potencia y linealidad). En x₀=1: pendiente 0, punto (−) tangencia (1,−2).</div></div>

```python
g = lambda x: x**3 - 3*x
gp = lambda x: 3*x**2 - 3  # g'(x)=3x²-3 derivada A MANO

x0_g = 1
x = np.linspace(-3, 3, 400)
tangente_g = g(x0_g) + gp(x0_g) * (x - x0_g)

plt.plot(x, g(x), color=INDIGO, lw=2.5, label=r"$g(x)=x^3-3x$")
plt.plot(x, tangente_g, color=CORAL, lw=2, ls="--", label=fr"tangente en $x_0={x0_g}$")
plt.scatter([x0_g], [g(x0_g)], color=PURPURA, s=70, zorder=5)
plt.axhline(0, color="#9CA3AF", lw=0.8); plt.axvline(0, color="#9CA3AF", lw=0.8)
plt.title(f"Pendiente de la tangente = g'({x0_g}) = {gp(x0_g)}", color=TINTA)
plt.legend(); plt.show()

print(f"g({x0_g}) = {g(x0_g)}, g'({x0_g}) = {gp(x0_g)} — tangente horizontal y=-2")
```

<div style="background:#ECFDF5;border:1px solid #A7F3D0;border-radius:10px;padding:14px 18px"><div style="font-weight:800;color:#047857;margin-bottom:6px">✓ Observación E2</div><div style="font-size:14px;color:#334155">La tangente en x₀=1 es horizontal (pendiente 0) y toca g(x) en (1,−2), un mínimo local. A diferencia de f(x)=x² donde la tangente cruza con pendiente 3, aquí la curva se aplana: la derivada cero marca punto crítico donde la función deja de crecer y empieza a decrecer (o viceversa).</div></div>

<h3 style="color:#1E293B;margin:14px 0 4px;font-size:17px">E3 · Descenso desde (−4, 3)</h3><p style="font-size:14px;color:#334155">Mismo tazón f(x,y)=x²+y², mismo α=0.1, 25 iteraciones, pero punto inicial <b>(−4, 3)</b> en lugar de (4,4). La trayectoria cambia de cuadrante pero converge al mismo mínimo (0,0).</p>

```python
# --- E3: mismo gradiente f3, nuevo punto inicial ---
punto_e3 = np.array([-4.0, 3.0])
alpha_e3 = 0.1
trayectoria_e3 = [punto_e3.copy()]

for _ in range(25):
    punto_e3 = punto_e3 - alpha_e3 * grad_f3(*punto_e3)
    trayectoria_e3.append(punto_e3.copy())

trayectoria_e3 = np.array(trayectoria_e3)
print("Primeros puntos E3:")
for i, p in enumerate(trayectoria_e3[:3]):
    print(f" x_{i} = ({p[0]:.3f}, {p[1]:.3f})")
print(f" ...\n x_25 = ({trayectoria_e3[-1,0]:.4f}, {trayectoria_e3[-1,1]:.4f})")

# Dibujar sobre el mismo mapa de contornos
xf = np.linspace(-5, 5, 200); yf = np.linspace(-5, 5, 200)
Xf, Yf = np.meshgrid(xf, yf); Zf = f3(Xf, Yf)

plt.contour(Xf, Yf, Zf, levels=20, cmap="viridis", alpha=0.6)
plt.plot(trayectoria_e3[:, 0], trayectoria_e3[:, 1], "o-", color=CORAL, ms=4, lw=1.5, label="trayectoria (−4,3)→mínimo")
plt.plot(trayectoria[:, 0], trayectoria[:, 1], "o-", color=INDIGO, ms=3, lw=1, alpha=0.5, label="trayectoria original (4,4)")
plt.scatter([0], [0], marker="*", s=260, color=PURPURA, zorder=5, label="mínimo")
plt.title("E3: dos trayectorias hacia el mismo mínimo", color=TINTA)
plt.xlabel("x"); plt.ylabel("y"); plt.legend(); plt.gca().set_aspect("equal")
plt.show()
```

<h3 style="color:#1E293B;margin:14px 0 4px;font-size:17px">E4 · Reflexión — ¿cómo sabe el modelo hacia dónde mover cada parámetro?</h3>

<div style="background:#F5F3FF;border:1px solid #DDD6FE;border-radius:12px;padding:18px 20px;font-size:14.5px;color:#334155;line-height:1.6"><p>El modelo no adivina: <b>calcula el gradiente de la función de pérdida</b> respecto a cada parámetro. Cada componente del gradiente es una derivada parcial que responde: si aumento un poco este parámetro, ¿cuánto sube o baja el error?</p><p>En las visualizaciones vimos que <b>∇f apunta cuesta arriba</b> (Viz 3, flechas perpendiculares a curvas de nivel) y que <b>−∇f apunta cuesta abajo</b>. El descenso hace <code>x_{k+1}=x_k−α·∇f(x_k)</code> (Viz 4): da un paso pequeño en dirección contraria al gradiente, porque ahí la pérdida disminuye más rápido. La magnitud del gradiente regula el paso — grande al inicio (pasa largo), pequeño cerca del mínimo (frena solo) — y α (tasa de aprendizaje) controla si avanza firme, lento o diverge.</p><p>Entrenar es repetir eso para miles de parámetros a la vez: en cada lote calcula el gradiente, mueve cada peso un poco contra su pendiente, y el error promedio va bajando hasta un mínimo. Por eso entender derivada y gradiente es entender el corazón del aprendizaje.</p></div>
